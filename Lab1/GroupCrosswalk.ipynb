## 1. Key differences

The detail sits in the tables below. These are the differences that actually matter.
The ones needing a single group answer are collected as a decision list in Section 9.

- **Target variable choice: three-class versus two-class.** Harvey collapses the target to
  binary, Diabetes vs Not Diabetes, and reports 13.9 percent. Sydney works with all three
  classes throughout. Wolfgang moves between both depending on the section. The headline
  prevalence number changes depending on which framing wins, so this is the first thing
  the group has to agree on.

- **Prediabetes is handled three ways.** Harvey reports it as likely unlearnable at 4,631
  examples. Sydney treats it as a full class alongside the others. Wolfgang holds it out of
  training entirely. This needs a decision, not a compromise.

- **BMI capping done three ways.** Harvey caps at the 99th percentile, which lands at
  BMI 50, and keeps the uncapped column alongside it. Wolfgang winsorises at BMI 70.
  Sydney describes the distribution but applies no cap. Three different treatments of the
  same tail.

- **Duplicates: same conclusion, different rigor.** All three keep the roughly 9 percent
  of rows that are exact duplicates. Harvey and Wolfgang both measured what dropping them
  would do and found prevalence rises because healthy profiles are deleted first. Sydney
  reasons correctly that duplicates need not be mistakes but does not quantify the effect.

- **Correlation method.** Harvey and Sydney both run Pearson correlation on coded ordinals.
  Wolfgang rejects that approach and uses Cramer's V with chi-square instead. His
  objection is methodologically right, since Pearson on an ordinal code only captures the
  linear part of the trend.

- **Only Wolfgang pulls in census data.** He takes 2015 American Community Survey figures
  from the Census API and compares the BRFSS age, income, and education distributions
  against the US population. Nothing in the other two notebooks benchmarks the sample
  against anything external.

- **Two preliminary models proposed.** Harvey fits logistic regression as a baseline proof
  of concept on a 70/30 split. Wolfgang tunes a gradient-boosting model and uses it to
  estimate undiagnosed cases. Sydney fits no model and stops at PCA.

- **Diagnosis versus disease.** Harvey and Wolfgang both build their analysis around the
  fact that the target records a diagnosis rather than the condition, so the undiagnosed
  sit in the negative class. Sydney defines the three classes without that caveat, which
  changes how the findings should be read.

- **Multicollinearity is checked once.** Wolfgang runs variance inflation factors and
  reports the condition number. Harvey reaches a similar conclusion informally by
  reporting correlated clusters. Sydney does not examine redundancy among predictors.

- **Negative results are reported once.** Harvey keeps a feature that failed, the lifestyle
  score, specifically to report that it does not work. Neither of the other notebooks
  reports something that was tried and abandoned.
